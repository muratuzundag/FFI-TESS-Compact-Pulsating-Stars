Install this repository in the notebook environment first. This notebook contains no precomputed photometry. It uses online services only when run. Read the README and validation status before scientific use.

```python
from ffi_tess.cli import main
TIC = 422526868  # usage example; replace with your target
```

## Coverage inventory

An archive failure must not be read as no FFI coverage. Inspect the manifest.

```python
status = main(["--tic", str(TIC), "--output", "results_inventory"])
print("Exit status:", status)
```

```python
from pathlib import Path
import json
folder = Path("results_inventory") / f"TIC_{TIC}"
manifest = json.loads((folder / "manifest.json").read_text())
print(manifest["coverage_status"])
print(manifest["coverage"])
print(manifest["errors"])
```

## Optional extraction

Install the extraction extra first. Set RUN_EXTRACTION only after reviewing coverage and correction assumptions. The default circular mask and standard correction are not guarantees of signal preservation.

```python
RUN_EXTRACTION = False
if RUN_EXTRACTION:
    status = main(["--tic", str(TIC), "--extract", "--output", "results_extraction"])
    print("Exit status:", status)
```

## Inspect saved figures

Candidate peaks are not automatic detections. Compare raw and selected photometry, and examine the aperture.

```python
from IPython.display import Image, display
folder = Path("results_extraction") / f"TIC_{TIC}"
for figure in sorted(folder.glob("*_diagnostics.png")):
    display(Image(filename=str(figure)))
```
